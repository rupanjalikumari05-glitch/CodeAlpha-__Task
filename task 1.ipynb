```python
import requests
from bs4 import BeautifulSoup

url = "https://www.scrapethissite.com/pages/forms/"
response = requests.get(url)

soup = BeautifulSoup(response.text, "html.parser")

rows = soup.find_all("tr", class_="team")

# empty list banate hain data store karne ke liye
data = []

for row in rows:
    team = row.find("td", class_="name").text.strip()
    year = row.find("td", class_="year").text.strip()
    wins = row.find("td", class_="wins").text.strip()
    losses = row.find("td", class_="losses").text.strip()
    gf = row.find("td", class_="gf").text.strip()
    ga = row.find("td", class_="ga").text.strip()

    data.append([team, year, wins, losses, gf, ga])

# first 5 rows check
for i in data[:5]:
    print(i)
```

```python
import pandas as pd

# columns define karte hain
columns = ["Team", "Year", "Wins", "Losses", "Goals_For", "Goals_Against"]

# data ko dataframe me convert
df = pd.DataFrame(data, columns=columns)

# first 5 rows check
print(df.head())
```

```python
df.to_csv("hockey_dataset.csv", index=False)

print("CSV file saved successfully!")
```
