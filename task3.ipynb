```python
import pandas as pd

df = pd.read_csv("hockey_dataset.csv")

# convert to numeric
df["Wins"] = df["Wins"].astype(int)
df["Losses"] = df["Losses"].astype(int)
df["Goals_For"] = df["Goals_For"].astype(int)
df["Goals_Against"] = df["Goals_Against"].astype(int)

print(df.head())
```

```python
import matplotlib.pyplot as plt

top = df.sort_values(by="Wins", ascending=False).head(10)

plt.bar(top["Team"], top["Wins"])
plt.xticks(rotation=90)
plt.title("Top 10 Teams by Wins")
plt.xlabel("Teams")
plt.ylabel("Wins")
plt.show()
```

```python
yearly = df.groupby("Year")["Wins"].mean().reset_index()

plt.plot(yearly["Year"], yearly["Wins"])
plt.title("Average Wins Trend Over Years")
plt.xlabel("Year")
plt.ylabel("Average Wins")
plt.show()
```

```python
plt.scatter(df["Goals_For"], df["Wins"])
plt.title("Goals vs Wins Relationship")
plt.xlabel("Goals For")
plt.ylabel("Wins")
plt.show()
```

```python
top5 = df.sort_values(by="Wins", ascending=False).head(5)

plt.pie(top5["Wins"], labels=top5["Team"], autopct="%1.1f%%")
plt.title("Top 5 Teams Win Share")
plt.show()
```

## Key Insights

1. Top performing teams were identified based on highest wins.
2. There is a positive relationship between goals scored and number of wins.
3. Year-wise analysis shows fluctuation in average team performance.
4. A few teams consistently outperform others across multiple years.
5. Some teams show poor performance indicating need for improvement.
